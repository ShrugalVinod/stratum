```python
# 4. LEVEL 2 — recompute H1 and H2 from the released derived data (every labelled commercial
#    flight segment with its satellite label, baselines and held-out STRATUM forecasts)
import subprocess
URL = "https://github.com/ShrugalVinod/stratum/releases/download/v1.0.0/stratum-release-1.0.0.zip"
if subprocess.run(["curl", "-sfLo", "release.zip", URL]).returncode != 0:
    print("The v1.0.0 release is not published yet; level 2 works once it is.")
else:
    subprocess.run(["unzip", "-q", "-o", "release.zip", "-d", "release"], check=True)
    r = subprocess.run(["uv", "run", "stratum", "rescore", "--release", "release/1.0.0"],
                       capture_output=True, text=True)
    print(r.stdout[-6000:], r.stderr[-2000:])
```

**Reading level 2:** the release holds commercial flights only (privacy), while the logged runs
used every labelled segment, so expect close, not identical, numbers. H3 and H4 need per-segment
contrail energy forcing from the physics model, which is not in the release (level 3).
